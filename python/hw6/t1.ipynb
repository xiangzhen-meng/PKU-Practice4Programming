```python
import pandas as pd
```

```python
dt = pd.read_excel('./交易统计/finance.xlsx')
```

```python
dt = dt.drop(['工号', '姓名', '时段', '柜台'], axis=1)
```

```python
dt['日期'] = pd.to_datetime(dt['日期'])
out = dt.groupby('日期').sum().reset_index().sort_values(by=['交易额'], ascending=True)
out.insert(2, '星期', out['日期'].dt.day_name())
out['交易额'] = out['交易额'].astype(int)
print(out.iloc[:3].to_string(index=False, header=False))
```
