```python
import pandas as pd

pd.set_option("display.unicode.east_asian_width", True)

scores = [["男", 108, 115, 97], ["女", 115, 87, 105], ["女", 100, 60, 130], ["男", 112, 80, 50]]
names = ["刘一哥", "王二姐", "张三妹", "李四弟"]
courses = ["性别", "语文", "数学", "英语"]
df = pd.DataFrame(data=scores, index=names, columns=courses)
print(df)
```
