```python
s = "<td>数据1</td><td>数据2</td>"

# 贪婪：从第一个 <td> 一路吃到最后一个 </td>
print(re.findall(r"<td>.*</td>", s))
# ['<td>数据1</td><td>数据2</td>']   ← 只得到一个整体，不是两个

# 非贪婪：碰到第一个 </td> 就停
print(re.findall(r"<td>.*?</td>", s))
# ['<td>数据1</td>', '<td>数据2</td>']   ← 正确得到两个
```

```python
def outer():
  count = 0
  def inner():
    nonlocal count
    count += 1
    return count
  return inner

f = outer()
print(f())  # 1
print(f())  # 2
print(f())  # 3
```

```python
class A:
    __p = 20                     # 私有类属性

    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y
        self.__q = 30            # 私有实例属性

a = A()
# print(a.__q)                   # AttributeError
print(a._A__q)                   # 30，改写后仍可访问
print(A._A__p)                   # 20
```

```python
class Student:
    def __init__(self, score=0):
        self._score = score

    @property
    def score(self):
        return self._score

    @score.setter
    def score(self, value):
        if value < 0 or value > 100:
            raise ValueError("分数必须在 0~100 之间")
        self._score = value

s = Student()
s.score = 95         # 正常
print(s.score)       # 95
s.score = 10        # 分数必须在 0~100 之间
```

```python
def h():
    print("hello")
    m = yield 5           # 此处接收 send 传入的值
    print("m =", m)
    d = yield 12
    print("d =", d)
    yield 11

c = h()
print(next(c))            # hello → 5
print(next(c))            # m = None → 12
print(next(c))            # d = None → 13
```
