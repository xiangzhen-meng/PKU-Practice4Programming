```python
import bs4
```

```python
soup = bs4.BeautifulSoup(
  open('Book_Page1.html', encoding='utf-8'),
  'html.parser'
)
```

```python
price = soup.findAll('p', attrs={'class': 'price_color'})
priceSum = 0.0
for x in price:
  priceSum += eval(x.text[1:])
print(priceSum)
```

```python
books = soup.findAll('h3')
maxlen = 0
maxlen_name = ''
for book in books:
  a = book.find('a')
  
  name = a.get('title')
  if maxlen < len(name):
    maxlen = len(name)
    maxlen_name = name
print(maxlen_name)  

nums = ['One', 'Two', 'Three', 'Four', 'Five']
starCount = 0
for i in range(5):
  stars = soup.findAll('p', attrs={'class': nums[i]})
  starCount += len(stars)
```
